```python
# identify sales amount outliers
Q1 = df["Sales_Amount"].quantile(0.25)
Q3=df["Sales_Amount"].quantile(0.75)
IQR = Q3-Q1
lower_limit = Q1-1.5*IQR
upper_limit=Q3-1.5*IQR
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Sales_Amount"] < lower_limit) |
    (df["Sales_Amount"] > upper_limit)
]

print(outliers)
```

```python
print("Number of Outliers:", len(outliers))
```

```python
#Sales Amount Outliers using Boxplot
plt.boxplot(df["Sales_Amount"]) #create boxplot

plt.title("Sales Amount - Outlier Detection")
plt.ylabel("Sales Amount")
plt.show()
```

```python
#Quantity Outliers Identify
Q1_quantity = df["Quantity"].quantile(0.25)
Q3_quantity = df["Quantity"].quantile(0.75)

IQR_quantity = Q3_quantity - Q1_quantity

lower_quantity = Q1_quantity - 1.5 * IQR_quantity
upper_quantity = Q3_quantity + 1.5 * IQR_quantity

print("Lower Limit:", lower_quantity)
print("Upper Limit:", upper_quantity)
```

```python
#actual Quantity outlier rows filter
quantity_outliers = df[
    (df["Quantity"] < lower_quantity) | # or condition
    (df["Quantity"] > upper_quantity)
]

print(quantity_outliers)

```

```python
# Quantity Outliers Boxplot
plt.boxplot(df["Quantity"])

plt.title("Quantity - Outlier Detection")
plt.ylabel("Quantity")
plt.show()
```

```python
## correlation Analysis
correlation = df["Quantity"].corr(df["Sales_Amount"])

print("Correlation:", correlation)
```

```python
# correlation matrix
correlation_matrix = df[["Quantity", "Sales_Amount"]].corr()

print(correlation_matrix)
```

```python
# interpret correlation
correlation = df["Quantity"].corr(df["Sales_Amount"])

print("Correlation between Quantity and Sales Amount:", correlation)

if correlation > 0:
    print("Positive relationship")
elif correlation < 0:
    print("Negative relationship")
else:
    print("No linear relationship")
```

```python
# GroupBy + Multiple Aggregations
product_summary = df.groupby("Product").agg(
    Total_Sales=("Sales_Amount", "sum"),
    Average_Sales=("Sales_Amount", "mean"),
    Maximum_Sales=("Sales_Amount", "max"),
    Total_Quantity=("Quantity", "sum")
)

print(product_summary)
```

```python
## pivot table
pivot_sales = pd.pivot_table(
    df,
    values="Sales_Amount",
    index="Region",
    columns="Product",
    aggfunc="sum"
)

print(pivot_sales)
```

```python
## ranking
product_ranking = df.groupby("Product")["Sales_Amount"].sum()

product_ranking = product_ranking.sort_values(ascending=False)

product_ranking = product_ranking.rank(
    method="dense",
    ascending=False
)

print(product_ranking)
```

```python
## percentage contribution
product_sales = df.groupby("Product")["Sales_Amount"].sum()

total_sales = product_sales.sum()

product_percentage = (product_sales / total_sales) * 100

product_percentage = product_percentage.sort_values(ascending=False)

print(product_percentage)
```

```python
# Sales Amount Ranges using pd.cut()
df["Sales_Category"] = pd.cut( # divides numerical values into ranges/categories
    df["Sales_Amount"],
    bins=[0, 5000, 10000, float("inf")],
    labels=["Low", "Medium", "High"]
)

print(df[["Sales_Amount", "Sales_Category"]].head())
```

```python
# Business-Level Combined Analysis
region_summary = df.groupby("Region").agg(
    Total_Sales=("Sales_Amount", "sum"),
    Average_Order_Value=("Sales_Amount", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

region_summary = region_summary.sort_values(
    "Total_Sales",
    ascending=False
)

print(region_summary)
```
